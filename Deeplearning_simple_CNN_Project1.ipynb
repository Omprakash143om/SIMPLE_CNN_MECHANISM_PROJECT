```python
import tensorflow as tf
import numpy as np
```

```python
x=[1,2,3,4,5]
y=[3,4,5,6,7]
```

```python
x = np.array(x)
y = np.array(y)
```

```python
layer1=tf.keras.layers.Dense(units=4,input_dim=1)
layer2=tf.keras.layers.Dense(units=32)
layer3=tf.keras.layers.Dense(units=32)
layer4=tf.keras.layers.Dense(units=64)
layer5=tf.keras.layers.Dense(units=64)
layer6=tf.keras.layers.Dense(units=128)
layer7=tf.keras.layers.Dense(units=64)
layer8=tf.keras.layers.Dense(units=64)
layer9=tf.keras.layers.Dense(units=1)
```

```python
model=tf.keras.Sequential()
```

```python
model.add(layer1)
model.add(layer2)
model.add(layer3)
model.add(layer4)
model.add(layer5)
model.add(layer6)
model.add(layer7)
model.add(layer8)
model.add(layer9)
```

```python
model.compile(optimizer='adam',loss='mean_squared_error',metrics=['mean_squared_error'])
```

```python
model.fit(x,y,epochs=10)
```

```python
model.predict(np.array([16]))
```
